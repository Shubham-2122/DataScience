```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
df = pd.read_csv("Task4_Employee_Training.csv")

df

df.head()
```

```python
df.shape
df.info()
df.isnull().sum()
```

```python
before_mean = df["Before_Training_Productivity"].mean()

print("Before Training Mean:", before_mean)
```

```python
after_mean = df["After_Training_Productivity"].mean()

print("After Training Mean:", after_mean)
```

```python
print("Before Training:", before_mean)
print("After Training:", after_mean)

print("Average Improvement:", after_mean - before_mean)
```

```python
#paired t-test 
t_statistic, p_value = stats.ttest_rel(
    df["After_Training_Productivity"],
    df["Before_Training_Productivity"],
    alternative="greater"
)

print("T-Statistic:", t_statistic)
print("P-Value:", p_value)
```

```python
# signifacnce Level (Alpha)

alpha = 0.05

print("Significance Level : ",alpha)
```

```python
if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("Training significantly improves employee productivity.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is not enough evidence to conclude that training improves productivity.")
```

```python
mean_values = [
    before_mean,
    after_mean
]

labels = [
    "Before Training",
    "After Training"
]

plt.figure(figsize=(7, 5))

plt.bar(labels, mean_values)

plt.title("Employee Productivity Before vs After Training")
plt.xlabel("Training Period")
plt.ylabel("Average Productivity")

plt.show()
```

```python
print("EMPLOYEE TRAINING HYPOTHESIS TESTING REPORT")

print("-------------------------------------------")

print("Number of Employees:", len(df))

print("Before Training Mean:", round(before_mean, 2))

print("After Training Mean:", round(after_mean, 2))

print("Average Improvement:", round(after_mean - before_mean, 2))

print("T-Statistic:", round(t_statistic, 4))

print("P-Value:", round(p_value, 6))

print("Significance Level:", alpha)

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: Training significantly improves productivity.")
else:
    print("Decision: Fail to Reject H0")
    print("Conclusion: Insufficient evidence of productivity improvement.")
```
